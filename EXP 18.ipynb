```python
import os

print("=" * 50)
print("PHYSICAL, LOGICAL AND SPARSE ACQUISITION")
print("=" * 50)

# Physical Acquisition (Simulation)
print("\n1. PHYSICAL ACQUISITION")
print("-" * 30)
print("Status : Success")
print("Description : Complete disk image acquired (Simulation)")
print("Includes : Operating System, Deleted Files, Free Space, File System")

# Logical Acquisition
print("\n2. LOGICAL ACQUISITION")
print("-" * 30)
print("Status : Success")
print("Description : Accessible files acquired")

files = os.listdir()

print("\nAccessible Files:")
for file in files:
    print("->", file)

# Sparse Acquisition
print("\n3. SPARSE ACQUISITION")
print("-" * 30)
print("Status : Success")
print("Description : Only important evidence collected")

keywords = ["pdf", "doc", "txt", "jpg", "png"]

print("\nImportant Files:")
found = False
for file in files:
    if any(file.lower().endswith(ext) for ext in keywords):
        print("->", file)
        found = True

if not found:
    print("No important evidence files found.")

print("\nAcquisition Process Completed Successfully!")
```

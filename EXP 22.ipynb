```python
import hashlib
from datetime import datetime

print("=" * 50)
print("BIT STREAM COPY VERIFICATION")
print("=" * 50)

# Simulated Original Disk Data
original_data = "Digital Forensics Evidence Data"

# Simulated Bit Stream Copy
copied_data = "Digital Forensics Evidence Data"

# Calculate SHA-256 Hashes
original_hash = hashlib.sha256(original_data.encode()).hexdigest()
copied_hash = hashlib.sha256(copied_data.encode()).hexdigest()

print("\nVerification Time :", datetime.now())

print("\nOriginal Evidence Hash:")
print(original_hash)

print("\nCopied Evidence Hash:")
print(copied_hash)

# Verification
print("\nVerification Result")
print("-" * 30)

if original_hash == copied_hash:
    print("Status : SUCCESS")
    print("Result : Bit Stream Copy Verified Successfully.")
    print("Integrity : Original and Copy are Identical.")
else:
    print("Status : FAILED")
    print("Result : Copy Verification Failed.")
    print("Integrity : Data has been Modified.")

print("\nProcess Completed Successfully!")
```

```python
print("=" * 60)
print("FILE SYSTEM FEATURE SIMULATION (FAT vs NTFS vs EXT)")
print("=" * 60)

print("\n1. FAT FILE SYSTEM")
print("-" * 30)
print("Status           : Supported")
print("Maximum File Size: 4 GB")
print("Security         : No File Permissions")
print("Journaling       : Not Available")
print("Operating System : Windows, USB Drives")

print("\n2. NTFS FILE SYSTEM")
print("-" * 30)
print("Status           : Supported")
print("Maximum File Size: 16 EB (Theoretical)")
print("Security         : File Permissions & Encryption")
print("Journaling       : Available")
print("Operating System : Windows")

print("\n3. EXT FILE SYSTEM")
print("-" * 30)
print("Status           : Supported")
print("Maximum File Size: Up to 16 TB (EXT4)")
print("Security         : Linux File Permissions")
print("Journaling       : Available")
print("Operating System : Linux")

print("\nCOMPARISON")
print("-" * 30)
print("FAT  -> Simple, Fast, No Security")
print("NTFS -> Secure, Journaling, Windows")
print("EXT  -> Secure, Journaling, Linux")

print("\nSimulation Completed Successfully!")
```

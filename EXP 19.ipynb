```python
from datetime import datetime
import platform

print("=" * 50)
print("LIVE VS DEAD ACQUISITION")
print("=" * 50)

# Live Acquisition
print("\n1. LIVE ACQUISITION")
print("-" * 30)
print("Status       : Success")
print("System State : Running")
print("Capture Time :", datetime.now())
print("Operating System :", platform.system(), platform.release())
print("Description  : Volatile data (RAM, running processes, network connections) can be collected while the system is ON.")

# Dead Acquisition
print("\n2. DEAD ACQUISITION")
print("-" * 30)
print("Status       : Success")
print("System State : Powered OFF")
print("Description  : Non-volatile data (Hard Disk, SSD, USB) is collected after shutting down the system.")
print("Captured Data: Files, Folders, Deleted Files, Disk Image")

# Comparison
print("\nCOMPARISON")
print("-" * 30)
print("Live Acquisition : Captures Volatile Evidence")
print("Dead Acquisition : Captures Non-Volatile Evidence")

print("\nAcquisition Demonstration Completed Successfully!")
```

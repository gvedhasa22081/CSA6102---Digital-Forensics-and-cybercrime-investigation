```python
print("=" * 50)
print("SLACK SPACE CALCULATION")
print("=" * 50)

# Example values
file_size = 3500       # File size in bytes
cluster_size = 4096    # Cluster size in bytes

# Calculate Slack Space
slack_space = cluster_size - (file_size % cluster_size)

# If file exactly fits a cluster
if file_size % cluster_size == 0:
    slack_space = 0

print("\nFile Size      :", file_size, "Bytes")
print("Cluster Size   :", cluster_size, "Bytes")
print("Slack Space    :", slack_space, "Bytes")

print("\nExplanation:")
print("Slack Space is the unused space left in the last allocated cluster of a file.")

print("\nCalculation:")
print("Slack Space = Cluster Size - (File Size % Cluster Size)")
print("Slack Space =", cluster_size, "-", (file_size % cluster_size), "=", slack_space)

print("\nStatus : SUCCESS")
print("Slack Space Calculated Successfully!")
```

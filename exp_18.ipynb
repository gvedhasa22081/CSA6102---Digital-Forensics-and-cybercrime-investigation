```python
import os
import hashlib
import csv
from datetime import datetime

# Folder containing digital evidence
EVIDENCE_FOLDER = "evidence"

# Output CSV file
OUTPUT_FILE = "evidence_catalog.csv"


# Function to calculate SHA-256 hash
def calculate_hash(file_path):
    sha256 = hashlib.sha256()

    with open(file_path, "rb") as file:
        while True:
            data = file.read(4096)
            if not data:
                break
            sha256.update(data)

    return sha256.hexdigest()


# Create evidence catalog
with open(OUTPUT_FILE, mode="w", newline="") as csvfile:
    writer = csv.writer(csvfile)

    # CSV Header
    writer.writerow([
        "Evidence ID",
        "File Name",
        "File Size (Bytes)",
        "Created Time",
        "Modified Time",
        "SHA-256 Hash"
    ])

    evidence_id = 1

    # Scan all files in evidence folder
    for filename in os.listdir(EVIDENCE_FOLDER):
        file_path = os.path.join(EVIDENCE_FOLDER, filename)

        if os.path.isfile(file_path):
            size = os.path.getsize(file_path)

            created = datetime.fromtimestamp(
                os.path.getctime(file_path)
            ).strftime("%Y-%m-%d %H:%M:%S")

            modified = datetime.fromtimestamp(
                os.path.getmtime(file_path)
            ).strftime("%Y-%m-%d %H:%M:%S")

            file_hash = calculate_hash(file_path)

            writer.writerow([
                evidence_id,
                filename,
                size,
                created,
                modified,
                file_hash
            ])

            print(f"[✓] Catalogued: {filename}")
            evidence_id += 1

print("\nDigital Evidence Catalog Created Successfully!")
print("Saved as:", OUTPUT_FILE)
```

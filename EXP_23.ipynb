```python
import hashlib

print("=" * 50)
print("HASHING & THE AVALANCHE EFFECT")
print("=" * 50)

# Original Data
data1 = "Digital Forensics"

# Slightly Modified Data (one character changed)
data2 = "Digital forensicS"

# Generate SHA-256 Hashes
hash1 = hashlib.sha256(data1.encode()).hexdigest()
hash2 = hashlib.sha256(data2.encode()).hexdigest()

print("\nOriginal Data:")
print(data1)

print("\nModified Data:")
print(data2)

print("\nSHA-256 Hash of Original Data:")
print(hash1)

print("\nSHA-256 Hash of Modified Data:")
print(hash2)

# Compare Hashes
print("\nVerification Result")
print("-" * 30)

if hash1 == hash2:
    print("Hashes are IDENTICAL")
else:
    print("Hashes are DIFFERENT")
    print("Avalanche Effect Demonstrated Successfully!")

print("\nExperiment Completed Successfully!")
```

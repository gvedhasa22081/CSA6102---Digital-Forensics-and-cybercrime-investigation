```python
from datetime import datetime

print("=" * 50)
print("IMAGING VS DUPLICATION")
print("=" * 50)

# Imaging
print("\n1. IMAGING")
print("-" * 30)
print("Status        : Success")
print("Operation     : Disk Imaging")
print("Capture Time  :", datetime.now())
print("Description   : Creates an exact bit-by-bit copy of the storage device.")
print("Includes      : Files, Deleted Files, Free Space, File System Metadata")
print("Output File   : disk_image.img")

# Duplication
print("\n2. DUPLICATION")
print("-" * 30)
print("Status        : Success")
print("Operation     : Disk Duplication")
print("Description   : Creates an exact copy of one storage device onto another.")
print("Target Device : Backup Drive")
print("Output        : Duplicate Disk Created")

# Comparison
print("\nCOMPARISON")
print("-" * 30)
print("Imaging    : Creates an image file (.img, .dd, .E01)")
print("Duplication: Creates another physical copy of the disk")
print("Imaging    : Used for forensic analysis")
print("Duplication: Used for backup and recovery")

print("\nOperation Completed Successfully!")
```
